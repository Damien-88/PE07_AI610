### Students will develop a Python program where:
#### You are tasked with optimizing the delivery route for a company that delivers packages to five locations. The company wants to minimize the total distance traveled. Use PSO to find the optimal sequence of visits to these locations.

```python
import numpy as np

# Delivery locations (x, y)
locations = np.array([[0, 0], [2, 3], [5, 1], [6, 4], [8, 0]])

# Distance function
def route_distance(route):
    dist = sum(np.linalg.norm(locations[route[i]] - locations[route[i+1]]) for i in range(len(route) - 1))
    return dist

# Particle Swarm Optimization
def PSO(num_particles=10, iterations=50):

    particles = []

    # Initialize swarm with random permutations of location indices
    for p in range(num_particles):
        route = np.random.permutation(len(locations))
        particles.append(route)

    # Each particle's own best route found so far
    personal_best = [particle.copy() for particle in particles]
    personal_best_distance = [route_distance(route) for route in personal_best]

    # Best route found across whole swarm
    best_index = np.argmin(personal_best_distance)
    global_best = personal_best[best_index].copy()
    global_best_distance = route_distance(global_best)

    for i in range(iterations):

        for j in range(num_particles):
            current_route = particles[j].copy()

            # Velocity toward personal best. 
            # Swap-based move pulls route one step closer to its own historical best.
            if np.random.random() < 0.5:

                for k in range(len(current_route)):

                    if current_route[k] != personal_best[j][k]:
                        target_index = np.where(current_route == personal_best[j][k])[0][0]
                        current_route[k], current_route[target_index] = current_route[target_index], current_route[k]

                        break

            # Velocity toward global best. 
            # Same swap idea but pulls toward swarm's best route.
            if np.random.random() < 0.5:

                for k in range(len(current_route)):

                    if current_route[k] != global_best[k]:
                        target_index = np.where(current_route == global_best[k])[0][0]
                        current_route[k], current_route[target_index] = current_route[target_index], current_route[k]

                        break

            particles[j] = current_route
            current_distance = route_distance(current_route)

            # Update personal best if move improved particle's own route
            if current_distance < personal_best_distance[j]:
                personal_best[j] = current_route.copy()
                personal_best_distance[j] = current_distance

            # Update global best if move improved on swarm's best route
            if current_distance < global_best_distance:
                global_best = current_route.copy()
                global_best_distance = current_distance

    return global_best

# Find and print the optimal route
optimal_route = PSO()
print("Optimal route:", optimal_route)
print("Total distance:", route_distance(optimal_route))
```
